Creating LangChain Vector Store (Backed by AstraDB)

```python
from langchain.embeddings import OpenAIEmbeddings
from langchain.vectorstores.cassandra import Cassandra

embeddings = OpenAIEmbeddings()

vectorstore = Cassandra.from_documents(
    documents=docs,
    embedding=embeddings,
    keyspace=None,         # ✅ Must be vector-enabled
    table_name="pdf_chunks",    # ✅ Will be auto-created
    session=None                # ✅ CassIO handles it
)
```

```python
from langchain.text_splitter import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=800, 
    chunk_overlap=200,
    length_function=len,
    )

texts = text_splitter.split_text(raw_text)
```

```python
texts[:40]
```

```python
vectorstore.add_texts(texts[:40])

print("Creating LangChain Vector Store (Backed by AstraDB)")
print("Inserted %i lines/chunks into the vector store." % len(texts[:40]))

astra_vector_Index =  VectorStoreIndexWrapper(
    vectorstore=vectorstore,
)
```

Run the Q&A Cycle

```python
first_question = True


while True:
    if first_question:
        question = input("Ask a question about the PDF: ").strip()
        first_question = False
    else:
        question = input("Ask another question (or type 'exit' to quit): ").strip()

    if question.lower() == 'exit':
        print("Exiting the Q&A cycle.")
        break

    if not question:
        print("Please enter a valid question.")
        continue

    first_question = False

    print("\n QUESTION: ", question)
    answer = astra_vector_Index.query(question, llm = llm, verbose=True).strip()
    print("Answering the question using LangChain and AstraDB...\n")
    print(f"Answer: {answer}")
    


    print("\n---\n")
    print("First Document by relevance:")
    for doc, score in vectorstore.similarity_search_with_score(question, k=1):
        # Print the score and the first 500 characters of the document content
        print(f"Score: {score}")
        print(f"Document: {doc.page_content[:500]}...\n")
```
