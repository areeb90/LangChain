## Combining Multiple Chains Using simple Sequential Chain


### Find the capital of the country and then its famous places to visit

```python
capital_template=PromptTemplate(input_variables=['country'],
template="Please tell me the capital of the {country}")

capital_chain=LLMChain(llm=llm,prompt=capital_template)



famous_template=PromptTemplate(input_variables=['capital'],
template="Suggest me some amazing places to visit in {capital}")
```

```python
famous_chain=LLMChain(llm=llm,prompt=famous_template)
```

```python
from langchain.chains import SimpleSequentialChain  

sequential_chain=SimpleSequentialChain(chains=[capital_chain,famous_chain],verbose=True)
print("Output 3")
print(sequential_chain.run("Pakistan"))
```

### Find the Biggest city of the country and the famous places to visit

```python
capital_template=PromptTemplate(input_variables=['country'],
template="Please tell me the biggest city of the {country}")

capital_chain=LLMChain(llm=llm,prompt=capital_template)



famous_template=PromptTemplate(input_variables=['biggest_city'],
template="Suggest me some amazing places to visit in {biggest_city}")
famous_chain=LLMChain(llm=llm,prompt=famous_template)
```

```python
from langchain.chains import SimpleSequentialChain

sequential_chain=SimpleSequentialChain(chains=[capital_chain,famous_chain],verbose=True)
print("Output 4")
print(sequential_chain.run("Pakistan"))
```

# 🧠 What Are chat_models in Langchain?
`chat_models` are LLMs designed to handle chat messages instead of plain text strings.

Unlike `llms.OpenAI (which uses raw strings)`, chat_models.ChatOpenAI uses structured messages — making it ideal for:

- Multi-turn conversations

- Agent frameworks

- Chain-of-thought prompting

- Better role separation (user/system/assistant)

# Prompt Template + LLM +Output Parsers


 ## PromptTemplate

A `PromptTemplate` is a template engine for generating prompts dynamically.
It lets you plug values into prompt strings — just like filling out a form.

💡 Why use it?
- Keeps your prompts clean and reusable

- Helps you standardize input to LLMs

- Works great with variables, chains, and agents



## OutputParser

An `OutputParser` takes the raw output from the LLM and converts it into structured data — like a Python dictionary, list, number, or JSON.

💡 Why use it?
- LLMs return plain text — but we often want structured data

- Output parsers help you extract meaning reliably

- Crucial in tools, agents, JSON workflows, etc.

```python
from langchain.chat_models import ChatOpenAI
from langchain.prompts.chat import ChatPromptTemplate, HumanMessagePromptTemplate, SystemMessagePromptTemplate
from langchain.schema import BaseOutputParser
```

### Comma seperated outputs

```python
class CommaSeparatedListOutputParser(BaseOutputParser):
    """Parse a comma-separated list of items."""
    def parse(self, text: str):
        return text.strip().split(", ")
```

```python
template = " You are a helpful coding geek. Please provide a comma-separated list of the following topics: {topics}"
human_template = "{text}"


chatPrompt = ChatPromptTemplate.from_messages(
    [ ("system", template),
    ("human", human_template)
])
```

```python
chain = chatPrompt | chatllm | CommaSeparatedListOutputParser()
```

```python
chain.invoke({"topics": "ML, DL, Generative AI", "text": "What are the skills needed for each domain?"})
```

```python
class showOutputInTableOutputParser(BaseOutputParser):
    """Parse the output and show it in a table format."""
    def parse(self, text: str):
        # Here you can implement logic to format the text into a table
        # spacing on every line
        # For simplicity, we will just return the text as is
        return text.strip()

```

```python
template = "You are a traveller guide, please provide a list of amazing places to visit in {country}. and specifically in that {province}"
human_template = "{text}"

chatPrompt = ChatPromptTemplate.from_messages(
    [ ("system", template),
    ("human", human_template)
])
chain = chatPrompt | chatllm | showOutputInTableOutputParser()
chain.invoke({"country": "Canada", "province": "Nova Scotia", "text": "What are the best places to visit?"})
```
