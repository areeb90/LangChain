```python
# Load and split PDF document
pdf_path = '../Chat_with_PDF_using_langChain_and_AstraDB/DSA.pdf'  # replace with your PDF path
loader = PyPDFLoader(pdf_path)
documents = loader.load()
```

```python
pdfReader = PyPDFLoader(pdf_path)
documents = pdfReader.load()
```

```python
from typing_extensions import Concatenate


raw_text = ''
for i, page in enumerate(documents):
    content = page.page_content if hasattr(page, 'page_content') else None
    # Check if content is None or empty
    if content is None or content.strip() == '':
        print(f"Skipping page {i + 1} due to no text content.")
        continue  # Skip pages with no text content
    # If content is None or empty, skip this page
    
    if content is None:
        continue  # Skip pages with no text content
    # Append the text content of the page to raw_text

    
    raw_text += content
    # Add a newline character to separate pages
    raw_text += f'\n\n--- Page {i + 1} ---\n\n'
    # If not the last page, add a separator
    if i < len(documents) - 1:
        raw_text += '\n\n--- End of Page ---\n\n'
```

```python
# Embed and store in AstraDB
embedding = OpenAIEmbeddings()
vstore = Cassandra(
    embedding=embedding,
    table_name="pdf_chunks",
    session=None,
    keyspace="pdfchat",
)
vstore.add_documents(docs)
```
